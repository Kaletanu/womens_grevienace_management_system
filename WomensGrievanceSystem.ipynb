```python
conn = sqlite3.connect("grievance.db")
cursor = conn.cursor()

# Users table
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    user_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT UNIQUE NOT NULL,
    mobile TEXT NOT NULL,
    password TEXT NOT NULL
)
""")

# Grievances table
cursor.execute("""
CREATE TABLE IF NOT EXISTS grievances (
    grievance_id TEXT PRIMARY KEY,
    user_id INTEGER,
    category TEXT NOT NULL,
    priority TEXT NOT NULL,
    subject TEXT NOT NULL,
    description TEXT NOT NULL,
    date TEXT NOT NULL,
    status TEXT NOT NULL,
    FOREIGN KEY(user_id) REFERENCES users(user_id)
)
""")

# Responses table
cursor.execute("""
CREATE TABLE IF NOT EXISTS responses (
    response_id INTEGER PRIMARY KEY AUTOINCREMENT,
    grievance_id TEXT,
    remark TEXT,
    response_date TEXT,
    FOREIGN KEY(grievance_id) REFERENCES grievances(grievance_id)
)
""")

conn.commit()
conn.close()

print("All database tables created successfully!")
```

```python
def registration_page():

    clear_page()

    tk.Label(
        content_frame,
        text="Create User Account",
        font=("Arial", 26, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=25)

    form = tk.Frame(
        content_frame,
        bg=WHITE,
        padx=45,
        pady=30,
        relief="solid",
        bd=1
    )

    form.pack()

    # Name
    tk.Label(
        form,
        text="Full Name",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=0, column=0, sticky="w", pady=10)

    name_entry = tk.Entry(
        form,
        width=35,
        font=("Arial", 12)
    )

    name_entry.grid(row=0, column=1, padx=25)

    # Email
    tk.Label(
        form,
        text="Email",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=1, column=0, sticky="w", pady=10)

    email_entry = tk.Entry(
        form,
        width=35,
        font=("Arial", 12)
    )

    email_entry.grid(row=1, column=1, padx=25)

    # Mobile
    tk.Label(
        form,
        text="Mobile Number",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=2, column=0, sticky="w", pady=10)

    mobile_entry = tk.Entry(
        form,
        width=35,
        font=("Arial", 12)
    )

    mobile_entry.grid(row=2, column=1, padx=25)

    # Password
    tk.Label(
        form,
        text="Password",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=3, column=0, sticky="w", pady=10)

    password_entry = tk.Entry(
        form,
        width=35,
        font=("Arial", 12),
        show="*"
    )

    password_entry.grid(row=3, column=1, padx=25)

    # Confirm Password
    tk.Label(
        form,
        text="Confirm Password",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=4, column=0, sticky="w", pady=10)

    confirm_entry = tk.Entry(
        form,
        width=35,
        font=("Arial", 12),
        show="*"
    )

    confirm_entry.grid(row=4, column=1, padx=25)

    def register_user():

        name = name_entry.get().strip()
        email = email_entry.get().strip()
        mobile = mobile_entry.get().strip()
        password = password_entry.get()
        confirm = confirm_entry.get()

        if name == "" or email == "" or mobile == "" or password == "":
            messagebox.showwarning(
                "Required",
                "Please fill all fields."
            )
            return

        if password != confirm:
            messagebox.showerror(
                "Error",
                "Passwords do not match."
            )
            return

        try:

            conn = sqlite3.connect("grievance.db")
            cursor = conn.cursor()

            cursor.execute("""
            INSERT INTO users
            (name, email, mobile, password)
            VALUES (?, ?, ?, ?)
            """, (name, email, mobile, password))

            conn.commit()
            conn.close()

            messagebox.showinfo(
                "Registration Successful",
                "Account created successfully!\n\nPlease login."
            )

            user_login()

        except sqlite3.IntegrityError:

            messagebox.showerror(
                "Error",
                "This email is already registered."
            )

    tk.Button(
        form,
        text="CREATE ACCOUNT",
        font=("Arial", 12, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=22,
        height=2,
        command=register_user
    ).grid(
        row=5,
        column=0,
        columnspan=2,
        pady=25
    )
```

```python
def user_login():

    clear_page()

    tk.Label(
        content_frame,
        text="User Login",
        font=("Arial", 28, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=35)

    login_frame = tk.Frame(
        content_frame,
        bg=WHITE,
        padx=50,
        pady=40,
        relief="solid",
        bd=1
    )

    login_frame.pack()

    tk.Label(
        login_frame,
        text="Email",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=0, column=0, pady=12, sticky="w")

    email_entry = tk.Entry(
        login_frame,
        width=35,
        font=("Arial", 12)
    )

    email_entry.grid(row=0, column=1, padx=20)

    tk.Label(
        login_frame,
        text="Password",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=1, column=0, pady=12, sticky="w")

    password_entry = tk.Entry(
        login_frame,
        width=35,
        font=("Arial", 12),
        show="*"
    )

    password_entry.grid(row=1, column=1, padx=20)

    def login():

        global current_user_id
        global current_user_name

        email = email_entry.get().strip()
        password = password_entry.get()

        conn = sqlite3.connect("grievance.db")
        cursor = conn.cursor()

        cursor.execute("""
        SELECT user_id, name
        FROM users
        WHERE email=? AND password=?
        """, (email, password))

        result = cursor.fetchone()

        conn.close()

        if result:

            current_user_id = result[0]
            current_user_name = result[1]

            messagebox.showinfo(
                "Login Successful",
                "Welcome " + current_user_name
            )

            user_dashboard()

        else:

            messagebox.showerror(
                "Login Failed",
                "Invalid email or password."
            )

    tk.Button(
        login_frame,
        text="LOGIN",
        font=("Arial", 12, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=20,
        height=2,
        command=login
    ).grid(
        row=2,
        column=0,
        columnspan=2,
        pady=25
    )

    tk.Button(
        content_frame,
        text="Create New Account",
        font=("Arial", 11),
        command=registration_page
    ).pack()
```

```python
def user_dashboard():

    clear_page()

    tk.Label(
        content_frame,
        text="User Dashboard",
        font=("Arial", 28, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=(30, 5))

    tk.Label(
        content_frame,
        text="Welcome, " + current_user_name,
        font=("Arial", 14),
        bg=BG_COLOR,
        fg=GRAY
    ).pack()

    # Statistics
    conn = sqlite3.connect("grievance.db")
    cursor = conn.cursor()

    cursor.execute("""
    SELECT COUNT(*)
    FROM grievances
    WHERE user_id=?
    """, (current_user_id,))

    total = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*)
    FROM grievances
    WHERE user_id=? AND status='Pending'
    """, (current_user_id,))

    pending = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*)
    FROM grievances
    WHERE user_id=? AND status='In Progress'
    """, (current_user_id,))

    progress = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*)
    FROM grievances
    WHERE user_id=? AND status='Resolved'
    """, (current_user_id,))

    resolved = cursor.fetchone()[0]

    conn.close()

    stats = tk.Frame(
        content_frame,
        bg=BG_COLOR
    )

    stats.pack(pady=30)

    def create_stat(parent, title, value, color, column):

        frame = tk.Frame(
            parent,
            bg=WHITE,
            width=180,
            height=110,
            relief="solid",
            bd=1
        )

        frame.grid(
            row=0,
            column=column,
            padx=10
        )

        frame.pack_propagate(False)

        tk.Label(
            frame,
            text=title,
            font=("Arial", 11, "bold"),
            bg=WHITE,
            fg=GRAY
        ).pack(pady=(15, 5))

        tk.Label(
            frame,
            text=str(value),
            font=("Arial", 24, "bold"),
            bg=WHITE,
            fg=color
        ).pack()

    create_stat(stats, "TOTAL", total, PRIMARY, 0)
    create_stat(stats, "PENDING", pending, ORANGE, 1)
    create_stat(stats, "IN PROGRESS", progress, BLUE, 2)
    create_stat(stats, "RESOLVED", resolved, GREEN, 3)

    # Buttons
    actions = tk.Frame(
        content_frame,
        bg=BG_COLOR
    )

    actions.pack(pady=25)

    tk.Button(
        actions,
        text="REGISTER NEW GRIEVANCE",
        font=("Arial", 12, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=25,
        height=2,
        command=grievance_page
    ).grid(row=0, column=0, padx=10, pady=10)

    tk.Button(
        actions,
        text="MY GRIEVANCES",
        font=("Arial", 12, "bold"),
        bg=SECONDARY,
        fg=WHITE,
        width=25,
        height=2,
        command=my_grievances
    ).grid(row=0, column=1, padx=10, pady=10)

    tk.Button(
        actions,
        text="TRACK GRIEVANCE",
        font=("Arial", 12, "bold"),
        bg=BLUE,
        fg=WHITE,
        width=25,
        height=2,
        command=track_page
    ).grid(row=1, column=0, padx=10, pady=10)

    tk.Button(
        actions,
        text="LOGOUT",
        font=("Arial", 12, "bold"),
        bg=RED,
        fg=WHITE,
        width=25,
        height=2,
        command=logout
    ).grid(row=1, column=1, padx=10, pady=10)
```

```python
def grievance_page():

    clear_page()

    tk.Label(
        content_frame,
        text="Register New Grievance",
        font=("Arial", 27, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=25)

    form = tk.Frame(
        content_frame,
        bg=WHITE,
        padx=45,
        pady=25,
        relief="solid",
        bd=1
    )

    form.pack()

    # Category
    tk.Label(
        form,
        text="Category",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=0, column=0, sticky="w", pady=10)

    category = ttk.Combobox(
        form,
        values=[
            "Workplace",
            "Education",
            "Harassment",
            "Safety",
            "Domestic",
            "Financial",
            "Other"
        ],
        width=33,
        state="readonly"
    )

    category.grid(row=0, column=1, padx=20)
    category.set("Select Category")

    # Priority
    tk.Label(
        form,
        text="Priority",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=1, column=0, sticky="w", pady=10)

    priority = ttk.Combobox(
        form,
        values=[
            "Low",
            "Medium",
            "High"
        ],
        width=33,
        state="readonly"
    )

    priority.grid(row=1, column=1, padx=20)
    priority.set("Select Priority")

    # Subject
    tk.Label(
        form,
        text="Subject",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=2, column=0, sticky="w", pady=10)

    subject = tk.Entry(
        form,
        width=35,
        font=("Arial", 12)
    )

    subject.grid(row=2, column=1, padx=20)

    # Description
    tk.Label(
        form,
        text="Description",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=3, column=0, sticky="nw", pady=10)

    description = tk.Text(
        form,
        width=35,
        height=7,
        font=("Arial", 11)
    )

    description.grid(row=3, column=1, padx=20)

    def submit_grievance():

        cat = category.get()
        pri = priority.get()
        sub = subject.get().strip()
        desc = description.get("1.0", tk.END).strip()

        if cat == "Select Category":
            messagebox.showwarning(
                "Required",
                "Please select a category."
            )
            return

        if pri == "Select Priority":
            messagebox.showwarning(
                "Required",
                "Please select priority."
            )
            return

        if sub == "" or desc == "":
            messagebox.showwarning(
                "Required",
                "Please enter subject and description."
            )
            return

        # Generate unique grievance ID
        grievance_id = "GRV" + datetime.now().strftime("%Y%m%d%H%M%S")

        date = datetime.now().strftime(
            "%d-%m-%Y %H:%M"
        )

        conn = sqlite3.connect("grievance.db")
        cursor = conn.cursor()

        cursor.execute("""
        INSERT INTO grievances
        (
            grievance_id,
            user_id,
            category,
            priority,
            subject,
            description,
            date,
            status
        )
        VALUES (?, ?, ?, ?, ?, ?, ?, ?)
        """, (
            grievance_id,
            current_user_id,
            cat,
            pri,
            sub,
            desc,
            date,
            "Pending"
        ))

        conn.commit()
        conn.close()

        messagebox.showinfo(
            "Grievance Submitted",
            "Your grievance has been submitted successfully!\n\n"
            "Grievance ID:\n" + grievance_id
        )

        user_dashboard()

    tk.Button(
        form,
        text="SUBMIT GRIEVANCE",
        font=("Arial", 12, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=23,
        height=2,
        command=submit_grievance
    ).grid(
        row=4,
        column=0,
        columnspan=2,
        pady=20
    )
```

```python
def my_grievances():

    clear_page()

    tk.Label(
        content_frame,
        text="My Grievances",
        font=("Arial", 27, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=25)

    table_frame = tk.Frame(
        content_frame,
        bg=WHITE
    )

    table_frame.pack(
        padx=30,
        pady=10,
        fill="both",
        expand=True
    )

    columns = (
        "ID",
        "Category",
        "Priority",
        "Subject",
        "Date",
        "Status"
    )

    table = ttk.Treeview(
        table_frame,
        columns=columns,
        show="headings"
    )

    for col in columns:

        table.heading(
            col,
            text=col
        )

        table.column(
            col,
            width=140
        )

    table.pack(
        fill="both",
        expand=True,
        padx=10,
        pady=10
    )

    conn = sqlite3.connect("grievance.db")
    cursor = conn.cursor()

    cursor.execute("""
    SELECT
        grievance_id,
        category,
        priority,
        subject,
        date,
        status
    FROM grievances
    WHERE user_id=?
    ORDER BY rowid DESC
    """, (current_user_id,))

    records = cursor.fetchall()

    conn.close()

    for record in records:

        table.insert(
            "",
            tk.END,
            values=record
        )

    tk.Button(
        content_frame,
        text="BACK TO DASHBOARD",
        font=("Arial", 11, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=22,
        command=user_dashboard
    ).pack(pady=15)
```

```python
def track_page():

    clear_page()

    tk.Label(
        content_frame,
        text="Track Grievance",
        font=("Arial", 27, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=30)

    search_frame = tk.Frame(
        content_frame,
        bg=WHITE,
        padx=35,
        pady=25,
        relief="solid",
        bd=1
    )

    search_frame.pack()

    tk.Label(
        search_frame,
        text="Enter Grievance ID",
        font=("Arial", 12, "bold"),
        bg=WHITE
    ).grid(row=0, column=0, padx=10)

    grievance_entry = tk.Entry(
        search_frame,
        width=30,
        font=("Arial", 12)
    )

    grievance_entry.grid(row=0, column=1, padx=10)

    result_frame = tk.Frame(
        content_frame,
        bg=BG_COLOR
    )

    result_frame.pack(
        pady=30,
        fill="both",
        expand=True
    )

    def search_grievance():

        for widget in result_frame.winfo_children():
            widget.destroy()

        grievance_id = grievance_entry.get().strip()

        if grievance_id == "":
            messagebox.showwarning(
                "Required",
                "Please enter Grievance ID."
            )
            return

        conn = sqlite3.connect("grievance.db")
        cursor = conn.cursor()

        cursor.execute("""
        SELECT
            category,
            priority,
            subject,
            description,
            date,
            status
        FROM grievances
        WHERE grievance_id=?
        """, (grievance_id,))

        result = cursor.fetchone()

        cursor.execute("""
        SELECT remark, response_date
        FROM responses
        WHERE grievance_id=?
        ORDER BY response_id DESC
        LIMIT 1
        """, (grievance_id,))

        response = cursor.fetchone()

        conn.close()

        if result is None:

            messagebox.showerror(
                "Not Found",
                "Grievance ID not found."
            )

            return

        category_value, priority_value, subject_value, description_value, date_value, status_value = result

        card = tk.Frame(
            result_frame,
            bg=WHITE,
            padx=35,
            pady=25,
            relief="solid",
            bd=1
        )

        card.pack(
            padx=100,
            fill="x"
        )

        tk.Label(
            card,
            text="Grievance ID: " + grievance_id,
            font=("Arial", 16, "bold"),
            bg=WHITE,
            fg=PRIMARY
        ).pack(anchor="w", pady=5)

        tk.Label(
            card,
            text="Category: " + category_value,
            font=("Arial", 12),
            bg=WHITE,
            fg=TEXT_COLOR
        ).pack(anchor="w", pady=3)

        tk.Label(
            card,
            text="Priority: " + priority_value,
            font=("Arial", 12),
            bg=WHITE,
            fg=TEXT_COLOR
        ).pack(anchor="w", pady=3)

        tk.Label(
            card,
            text="Subject: " + subject_value,
            font=("Arial", 12),
            bg=WHITE,
            fg=TEXT_COLOR
        ).pack(anchor="w", pady=3)

        tk.Label(
            card,
            text="Date: " + date_value,
            font=("Arial", 12),
            bg=WHITE,
            fg=TEXT_COLOR
        ).pack(anchor="w", pady=3)

        tk.Label(
            card,
            text="STATUS: " + status_value,
            font=("Arial", 15, "bold"),
            bg=WHITE,
            fg=(
                GREEN if status_value == "Resolved"
                else ORANGE if status_value == "Pending"
                else BLUE
            )
        ).pack(anchor="w", pady=10)

        remark_text = "No admin remark available."

        if response:
            remark_text = response[0]

        tk.Label(
            card,
            text="Admin Remark:",
            font=("Arial", 12, "bold"),
            bg=WHITE,
            fg=PRIMARY
        ).pack(anchor="w", pady=(10, 3))

        tk.Label(
            card,
            text=remark_text,
            font=("Arial", 11),
            bg=WHITE,
            fg=TEXT_COLOR,
            wraplength=800,
            justify="left"
        ).pack(anchor="w")

    tk.Button(
        search_frame,
        text="SEARCH",
        font=("Arial", 11, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=12,
        command=search_grievance
    ).grid(row=0, column=2, padx=10)
```

```python
def admin_dashboard():

    clear_page()

    tk.Label(
        content_frame,
        text="Administrator Dashboard",
        font=("Arial", 28, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=20)

    # Statistics
    conn = sqlite3.connect("grievance.db")
    cursor = conn.cursor()

    cursor.execute("SELECT COUNT(*) FROM grievances")
    total = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*) FROM grievances
    WHERE status='Pending'
    """)
    pending = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*) FROM grievances
    WHERE status='In Progress'
    """)
    progress = cursor.fetchone()[0]

    cursor.execute("""
    SELECT COUNT(*) FROM grievances
    WHERE status='Resolved'
    """)
    resolved = cursor.fetchone()[0]

    conn.close()

    stats = tk.Frame(
        content_frame,
        bg=BG_COLOR
    )

    stats.pack(pady=10)

    def stat_box(title, value, color, column):

        box = tk.Frame(
            stats,
            bg=WHITE,
            width=170,
            height=90,
            relief="solid",
            bd=1
        )

        box.grid(
            row=0,
            column=column,
            padx=8
        )

        box.pack_propagate(False)

        tk.Label(
            box,
            text=title,
            font=("Arial", 10, "bold"),
            bg=WHITE,
            fg=GRAY
        ).pack(pady=(12, 2))

        tk.Label(
            box,
            text=str(value),
            font=("Arial", 22, "bold"),
            bg=WHITE,
            fg=color
        ).pack()

    stat_box("TOTAL", total, PRIMARY, 0)
    stat_box("PENDING", pending, ORANGE, 1)
    stat_box("IN PROGRESS", progress, BLUE, 2)
    stat_box("RESOLVED", resolved, GREEN, 3)

    # Table
    table_frame = tk.Frame(
        content_frame,
        bg=WHITE
    )

    table_frame.pack(
        padx=25,
        pady=20,
        fill="both",
        expand=True
    )

    columns = (
        "ID",
        "User ID",
        "Category",
        "Priority",
        "Subject",
        "Date",
        "Status"
    )

    table = ttk.Treeview(
        table_frame,
        columns=columns,
        show="headings"
    )

    for col in columns:

        table.heading(
            col,
            text=col
        )

        table.column(
            col,
            width=120
        )

    table.pack(
        fill="both",
        expand=True,
        padx=10,
        pady=10
    )

    conn = sqlite3.connect("grievance.db")
    cursor = conn.cursor()

    cursor.execute("""
    SELECT
        grievance_id,
        user_id,
        category,
        priority,
        subject,
        date,
        status
    FROM grievances
    ORDER BY rowid DESC
    """)

    records = cursor.fetchall()

    conn.close()

    for record in records:

        table.insert(
            "",
            tk.END,
            values=record
        )

    # Buttons
    button_frame = tk.Frame(
        content_frame,
        bg=BG_COLOR
    )

    button_frame.pack(pady=10)

    tk.Button(
        button_frame,
        text="UPDATE SELECTED GRIEVANCE",
        font=("Arial", 11, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=25,
        command=lambda: update_grievance(table)
    ).grid(row=0, column=0, padx=10)

    tk.Button(
        button_frame,
        text="REFRESH",
        font=("Arial", 11, "bold"),
        bg=BLUE,
        fg=WHITE,
        width=15,
        command=admin_dashboard
    ).grid(row=0, column=1, padx=10)

    tk.Button(
        button_frame,
        text="LOGOUT",
        font=("Arial", 11, "bold"),
        bg=RED,
        fg=WHITE,
        width=15,
        command=show_home
    ).grid(row=0, column=2, padx=10)
```

```python
def update_grievance(table):

    selected = table.selection()

    if not selected:

        messagebox.showwarning(
            "Select Grievance",
            "Please select a grievance from the table."
        )

        return

    values = table.item(
        selected[0],
        "values"
    )

    grievance_id = values[0]

    update_window = tk.Toplevel(root)

    update_window.title(
        "Update Grievance"
    )

    update_window.geometry(
        "500x400"
    )

    update_window.configure(
        bg=BG_COLOR
    )

    tk.Label(
        update_window,
        text="Update Grievance",
        font=("Arial", 22, "bold"),
        bg=BG_COLOR,
        fg=PRIMARY
    ).pack(pady=25)

    tk.Label(
        update_window,
        text="Grievance ID: " + grievance_id,
        font=("Arial", 12, "bold"),
        bg=BG_COLOR
    ).pack(pady=10)

    tk.Label(
        update_window,
        text="Status",
        font=("Arial", 12, "bold"),
        bg=BG_COLOR
    ).pack(pady=5)

    status = ttk.Combobox(
        update_window,
        values=[
            "Pending",
            "In Progress",
            "Resolved"
        ],
        state="readonly",
        width=30
    )

    status.pack()

    status.set(values[6])

    tk.Label(
        update_window,
        text="Admin Remark",
        font=("Arial", 12, "bold"),
        bg=BG_COLOR
    ).pack(pady=(20, 5))

    remark = tk.Text(
        update_window,
        width=40,
        height=6,
        font=("Arial", 11)
    )

    remark.pack()

    def save_update():

        new_status = status.get()
        new_remark = remark.get(
            "1.0",
            tk.END
        ).strip()

        if new_remark == "":

            messagebox.showwarning(
                "Required",
                "Please enter an admin remark."
            )

            return

        conn = sqlite3.connect(
            "grievance.db"
        )

        cursor = conn.cursor()

        # Update status
        cursor.execute("""
        UPDATE grievances
        SET status=?
        WHERE grievance_id=?
        """, (
            new_status,
            grievance_id
        ))

        # Save response
        response_date = datetime.now().strftime(
            "%d-%m-%Y %H:%M"
        )

        cursor.execute("""
        INSERT INTO responses
        (
            grievance_id,
            remark,
            response_date
        )
        VALUES (?, ?, ?)
        """, (
            grievance_id,
            new_remark,
            response_date
        ))

        conn.commit()
        conn.close()

        messagebox.showinfo(
            "Updated",
            "Grievance updated successfully."
        )

        update_window.destroy()

        admin_dashboard()

    tk.Button(
        update_window,
        text="SAVE UPDATE",
        font=("Arial", 12, "bold"),
        bg=PRIMARY,
        fg=WHITE,
        width=20,
        height=2,
        command=save_update
    ).pack(pady=20)
```
